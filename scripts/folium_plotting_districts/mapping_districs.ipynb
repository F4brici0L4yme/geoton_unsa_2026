```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
```

```python
df = pd.read_csv('../../data/cleaned/districts_w_coords.csv')
df = df.dropna(subset=['LATITUD', 'LONGITUD'])
```

```python
centro = [df['LATITUD'].mean(), df['LONGITUD'].mean()]
m = folium.Map(location=centro, zoom_start=8, tiles='CartoDB positron')

cluster = MarkerCluster().add_to(m)

for _, row in df.iterrows():
    popup_html = f"""
    <b>{row['CEN_EDU']}</b><br>
    Distrito: {row['DISTRITO']}<br>
    Localidad: {row['LOCALIDAD']}<br>
    Cód. Mod.: {row['COD_MOD']}
    """
    folium.Marker(
        location=[row['LATITUD'], row['LONGITUD']],
        popup=folium.Popup(popup_html, max_width=250),
        tooltip=str(row['CEN_EDU']),
    ).add_to(cluster)

m.save('mapa_escuelas.html')
m
```
