# 📊 Adding Coords

**Objective:** Add LATITUD & LONGITUD COLUMNS

**Data Source:**  https://www2.congreso.gob.pe/sicr/cendocbib/con4_uibd.nsf/D4C6C1DE2CC033F205257A8700657161/$FILE/Arequipa_Poblaci%C3%B3n.pdf 

**Output:** districts_w_coords.csv

### Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# from dbfread import DBF  # descomentar si usas .dbf

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
pd.set_option('display.float_format', '{:.2f}'.format)
```

### Paths

```python
# Entradas
PATH_RAW_FULL_COORDS = '../../data/external/congreso_del_peru/distritos_arequipa.csv'
PATH_RAW_FULL_DISTRICTS = '../../data/cleaned/districts.csv'

# Salidas
PATH_OUT_FULL    = '../../data/cleaned/districts_w_coords.csv'
```

### Data Loading

```python
# --- Opción A: CSV ---
df_districts = pd.read_csv(PATH_RAW_FULL_DISTRICTS)
df_coords = pd.read_csv(PATH_RAW_FULL_COORDS)

# --- Opción B: DBF ---
# from dbfread import DBF
# df_a = pd.DataFrame(iter(DBF(PATH_RAW_A, encoding='cp1252')))
# df_b = pd.DataFrame(iter(DBF(PATH_RAW_B, encoding='utf-8')))

# --- Opción C: Excel ---
# df_a = pd.read_excel(PATH_RAW_A, sheet_name=0)
```

### Data Inspection

```python
lista_distritos = list(df_districts['DISTRITO'].unique())
lista_distritos_coords = list(df_coords['DISTRITO'].unique())
# ver si hay distritos en coords que no estén en districts
set(lista_distritos) - set(lista_distritos_coords)
```

### Filter

```python
#añadir columna de LATITUD y LONGITUD de df_coords a df_districts
# crear un diccionario con los distritos de coords como claves y las coordenadas como valores
coords_dict = dict(zip(df_coords['DISTRITO'], zip(df_coords['LATITUD'], df_coords['LONGITUD'])))
# crear columnas de LATITUD y LONGITUD en df_districts usando el diccionario
df_districts['LATITUD'] = df_districts['DISTRITO'].map(lambda x: coords_dict.get(x, (np.nan, np.nan))[0])
df_districts['LONGITUD'] = df_districts['DISTRITO'].map(lambda x: coords_dict.get(x, (np.nan, np.nan))[1])
```

```python
#ordenar por distrito alfabeticamente
df_districts = df_districts.sort_values(by='DISTRITO').reset_index(drop=True)
df_districts.head(10)
```

### Output

```python
# Guardar por separado
df_districts.to_csv(PATH_OUT_FULL, index=False)

print(f'Exportado: {PATH_OUT_FULL} ({df_districts.shape[0]} filas)')
```
