```python
from difflib import get_close_matches

CORRECCIONES = {
    "JOSE LUIS BUSTAMENTE Y RIVERO":        "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSE LUIS BUSTAMANTE Y RIV ERO":       "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSSE LUIS BUSTAMANTE Y RIVERO":       "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSÉ LUIS BUSTAMANTE Y RIVERO":        "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSÉ LUIS BUSTAMATE Y RIVERO":         "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSE LUIS BUSTAMANTE Y RIBERO":        "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSE LUIS BUSTAMENTE Y RIVERO":        "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSE LUIS BUSTAMANTE Y RIVERP":        "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSE LUIS BUSTAMATE Y RIVERO":         "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSE LUIS BUSTAMENTE Y RIVERO":        "JOSE LUIS BUSTAMANTE Y RIVERO",
    "JOSE LUIS BUSTAMANTE Y RIVERO":        "JOSE LUIS BUSTAMANTE Y RIVERO",  # canónico
    "JOSE LUIS BUSTAMENTE Y RIVERO":        "JOSE LUIS BUSTAMANTE Y RIVERO",

    # CERRO COLORADO
    "CERRO COLRADO":                        "CERRO COLORADO",
    "CERROCOLORADO":                        "CERRO COLORADO",
    "CERRO C0LORADO":                       "CERRO COLORADO",   # cero en lugar de O

    # ALTO SELVA ALEGRE
    "ALTO SELVA AELGRE":                    "ALTO SELVA ALEGRE",
    "ALTO SELVA LEGRE":                     "ALTO SELVA ALEGRE",
    "ALTO SELVA ALGRE":                     "ALTO SELVA ALEGRE",
    "ALTO  SELVA  ALEGRE":                  "ALTO SELVA ALEGRE",  # espacios dobles
    "ALTOS SELVA ALEGRE":                   "ALTO SELVA ALEGRE",
    "ALTO SELA ALEGRE":                     "ALTO SELVA ALEGRE",
    "SELVA ALEGRE":                         "ALTO SELVA ALEGRE",

    # MARIANO MELGAR
    "MARIANO MELGA":                        "MARIANO MELGAR",
    "MARIANO MELGARX":                      "MARIANO MELGAR",
    "MARIANOMELGAR":                        "MARIANO MELGAR",

    # MARIANO NICOLAS VALCARCEL
    "MARIANO NICOLAS VARCALCEL":            "MARIANO NICOLAS VALCARCEL",
    "MARIANO NICOLÁS VALCÁRCEL":            "MARIANO NICOLAS VALCARCEL",

    # JACOBO HUNTER
    "JACOBO DICKSON HUNTER":                "JACOBO HUNTER",
    "JACOBO  HUNTER":                       "JACOBO HUNTER",   # doble espacio

    # PAUCARPATA
    "PUACARPATA":                           "PAUCARPATA",
    "PAURCAPATA":                           "PAUCARPATA",
    "PAUCARAPATA":                          "PAUCARPATA",

    # ORCOPAMPA
    "ORCOPMPA":                             "ORCOPAMPA",
    "ORCOPAPA":                             "ORCOPAMPA",

    # SOCABAYA
    "SOCABAY":                              "SOCABAYA",
    "AMPLIACION SOCABAYA":                  "SOCABAYA",
    "SACABAYA":                             "SOCABAYA",

    # CALLALLI
    "CALLALI":                              "CALLALLI",

    # CABANACONDE
    "CABANCONDE":                           "CABANACONDE",

    # AREQUIPA
    "AREUIPA":                              "AREQUIPA",
    "CERCADO DE AREQUIPA":                  "CERCADO",

    # CARAVELI
    "CARAVELÍ":                             "CARAVELI",

    # CHAPARRA
    "CHÁPARRA":                             "CHAPARRA",

    # LA JOYA
    "JA JOYA":                              "LA JOYA",

    # NICOLAS DE PIEROLA
    "NICOLÁS DE PIÉROLA":                   "NICOLAS DE PIEROLA",

    # VICTOR / VITOR
    "VICTOR":                               "VITOR",
    "VÍTOR":                                "VITOR",

    # SABANDIA
    "SABANDÍA":                             "SABANDIA",

    # SAMUEL PASTOR
    "SAMUEL PASTOR LA PAMPA":               "SAMUEL PASTOR",

    # ACARI
    "ACARÍ":                                "ACARI",

    # MEJIA
    "MEJÍA":                                "MEJIA",

    # BELLA UNION (normalizar tilde)
    "BELLA UNIÓN":                          "BELLA UNION",

    # CAMANA (normalizar tilde)
    "CAMANÁ":                               "CAMANA",

    # RIO GRANDE (normalizar tilde)
    "RÍO GRANDE":                           "RIO GRANDE",

    # PUNTA DE BOMBON (normalizar tilde)
    "PUNTA DE BOMBÓN":                      "PUNTA DE BOMBON",

    # CHILCAIMARCA
    "CHILCAIMARCA":                         "CHILCAYMARCA",

    # DEAN VALDIVIA
    "DEAN VALDIVA":                         "DEAN VALDIVIA",

    # COCACHACRA
    "COCCHACRA":                            "COCACHACRA",

    # ANDAGUA
    "ANDAGUS":                              "ANDAGUA",

    # MACHAHUAY / MACHAGUAY (misma localidad)
    "MACHAHUAY":                            "MACHAGUAY",

    # CHARACATO
    "CHARACAO":                             "CHARACATO",

    # Entradas sin correspondencia geográfica reconocida
    # Se dejan mapeadas a valores más descriptivos o se normalizan
    "CALLE COLO SIN NUMERO":               "SIN DATOS",
    "MUNDO ECOLOGICO":                     "SIN DATOS",
    "JULIO ERNESTO PORTUGAL ESCOBEDO":     "SIN DATOS",
    "PEDREGAL":                            "MAJES",          # Pedregal es la ciudad capital del distrito de Majes
}


# ─────────────────────────────────────────────
# 2. LISTA CANÓNICA DE DISTRITOS
#    (los que tienen conteo >= 5 en el dataset original)
# ─────────────────────────────────────────────
DISTRITOS_CANONICOS = [
    "CERRO COLORADO", "AREQUIPA", "PAUCARPATA", "MAJES", "CAYMA",
    "SOCABAYA", "MARIANO MELGAR", "JOSE LUIS BUSTAMANTE Y RIVERO",
    "ALTO SELVA ALEGRE", "LA JOYA", "MIRAFLORES", "YURA",
    "JACOBO HUNTER", "SACHACA", "MOLLENDO", "YANAHUARA", "TIABAYA",
    "APLAO", "MARIANO NICOLAS VALCARCEL", "CERCADO", "URACA",
    "SAMUEL PASTOR", "CAMANA", "COCACHACRA", "CAYLLOMA", "PAMPAMARCA",
    "HUAYNACOTAS", "ORCOPAMPA", "CHARACATO", "CAYARANI", "UCHUMAYO",
    "YANAQUIHUA", "CHUQUIBAMBA", "PUYCA", "OCOÑA", "ACARI", "TISCO",
    "HUNTER", "TORO", "DEAN VALDIVIA", "CHACHAS", "CHIGUATA", "CHOCO",
    "COTAHUASI", "VIRACO", "NICOLAS DE PIEROLA", "CHAPARRA",
    "SAN JUAN DE TARUCANI", "QUICACHA", "ATICO", "ALCA", "CHIVAY",
    "TAPAY", "LLUTA", "CHALA", "RIO GRANDE", "SALAMANCA", "CHICHAS",
    "SANTA RITA DE SIGUAS", "SANTA ISABEL DE SIGUAS", "POLOBAYA",
    "VITOR", "HUANUHUANU", "ANDARAY", "SAYLA", "ISLAY", "TIPAN",
    "PUNTA DE BOMBON", "CAHUACHO", "PAMPACOLCA", "ANDAGUA", "CALLALLI",
    "YANQUE", "TOMEPAMPA", "SAN JUAN DE SIGUAS", "SAN ANTONIO DE CHUCA",
    "HUANCA", "CABANACONDE", "COPORAQUE", "MARISCAL CACERES", "ATIQUIPA",
    "SIBAYO", "MACHAGUAY", "YARABAMBA", "QUECHUALLA", "IRAY",
    "JOSE MARIA QUIMPER", "POCSI", "MOLLEBAYA", "AYO", "ACHOMA",
    "CARAVELI", "HUAMBO", "JAQUI", "CHARCANA", "HUANCARQUI",
    "CHILCAYMARCA", "YAUCA", "TUTI", "UÑON", "TAURIA", "LOMAS",
    "ICHUPAMPA", "SABANDIA", "MADRIGAL", "MACA", "LARI", "QUILCA",
    "BELLA UNION", "MEJIA", "QUEQUEÑA",
]


def curar_distrito(nombre: str) -> str:
    """
    Aplica correcciones en dos pasos:
      1. Mapa explícito (cubre todos los casos conocidos).
      2. Fuzzy-matching contra la lista canónica (fallback).
    """
    if pd.isna(nombre):
        return nombre

    # Normalizar espacios extras
    nombre_norm = " ".join(str(nombre).upper().split())

    # Paso 1: corrección explícita
    if nombre_norm in CORRECCIONES:
        return CORRECCIONES[nombre_norm]

    # Paso 2: si ya está en la lista canónica, no tocar
    if nombre_norm in DISTRITOS_CANONICOS:
        return nombre_norm

    # Paso 3: fuzzy-matching (umbral 0.82 para evitar falsos positivos)
    matches = get_close_matches(nombre_norm, DISTRITOS_CANONICOS, n=1, cutoff=0.82)
    if matches:
        return matches[0]

    # Sin match → devolver tal como está (para revisión manual)
    return nombre_norm


# ─────────────────────────────────────────────
# 3. APLICAR AL DATAFRAME
# ─────────────────────────────────────────────
# df_p_s["DISTRITO"] = df_p_s["DISTRITO"].apply(curar_distrito)

# ─────────────────────────────────────────────
# 4. VERIFICACIÓN: mostrar cambios realizados
# ─────────────────────────────────────────────
def mostrar_cambios(df: pd.DataFrame, columna: str = "DISTRITO") -> pd.DataFrame:
    """Devuelve un DataFrame con las filas que fueron modificadas."""
    original = df[columna].str.upper().str.strip()
    corregida = df[columna]
    mask = original != corregida
    return pd.DataFrame({
        "indice": df.index[mask],
        "original": original[mask].values,
        "corregido": corregida[mask].values,
    })


# ─────────────────────────────────────────────
# USO COMPLETO (descomentar para ejecutar):
# ─────────────────────────────────────────────
df_p_s["DISTRITO"] = df_p_s["DISTRITO"].apply(curar_distrito)
cambios = mostrar_cambios(df_p_s)
print(f"\nTotal de filas corregidas: {len(cambios)}")
print(cambios.to_string(index=False))
print("\nConteo final:")
for d, c in df_p_s["DISTRITO"].value_counts().items():
    print(f"  {d}: {c}")
```
