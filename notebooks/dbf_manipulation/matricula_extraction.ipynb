```python
import pandas as pd
from dbfread import DBF
```

```python
# 1. Cargar el archivo DBF
# Nota: Si tienes problemas con caracteres especiales, añade el parámetro encoding='latin1' o 'cp1252'
tabla_dbf = DBF('../../data/raw/censo_educativo/Matricula_01.dbf', encoding='utf-8')
```

```python
df = pd.DataFrame(iter(tabla_dbf))
df.head()
```

```python
#CAMBIOS PARA SOLO SACAR COLEGIOS DE AREQUIPA
#ver cantidad filas de colegios con nroced = 3ap
cantidad_primaria = df[df["NROCED"] == "3AP"].shape[0]
cantidad_secundaria = df[df["NROCED"] == "3AS"].shape[0]
print(cantidad_primaria, cantidad_secundaria)
#filtrar por colegios de arequipa usando CODGEO, debe de empezar por 04, lo demás puede variar
colegios_arequipa = df[df["CODGEO"].str.startswith("04")].shape[0]
print(colegios_arequipa)
```
