```python
import pandas as pd
from dbfread import DBF
```

```python
tabla_dbf_primaria = DBF('../../data/raw/censo_educativo/Lineal_3AP_1.dbf', encoding='cp1252')
tabla_dbf_secundaria = DBF('../../data/raw/censo_educativo/Lineal_3AS_1.dbf', encoding='cp1252')
df_listado_ieee = pd.read_html('../../data/raw/mapa_servicio_educativo/listado_iiee.xls')
df_matricula_suma = pd.read_csv('../../data/raw/censo_educativo/matricula_suma.csv')
```

```python
df_listado_ieee = pd.read_html('../../data/raw/mapa_servicio_educativo/listado_iiee.xls')
```

```python
df_primaria = pd.DataFrame(iter(tabla_dbf_primaria))
df_secundaria = pd.DataFrame(iter(tabla_dbf_secundaria))
```

```python
df_primaria.head()
```

```python
df_secundaria.head()
```

```python
#combinar df primaria y secundaria y solo dejar las columnas COD_MOD, CODLOCAL, CODGEO, CEN_EDU, DISTRITO, LOCALIDAD
df_primaria_secundaria = pd.concat([df_primaria, df_secundaria], ignore_index=True)
df_primaria_secundaria = df_primaria_secundaria[["COD_MOD", "CODLOCAL", "CODGEO", "CEN_EDU", "DISTRITO", "LOCALIDAD"]]
df_primaria_secundaria.head()
df_primaria_secundaria.to_csv('../../data/cleaned/primaria_secundaria.csv', index=False)
```
